# HPC uzdevumi

1. Izveidot mapi ar nosaukumu [vārds_uzvārds], piemēram:
    ```bash
    mkdir janis_berzins
    ```
2. Ieiet tajā mapē:
    ```bash
    cd janis_berzins
    ```
3. Izdrukāt mapes nosaukumu:
    ```bash
    pwd
    ```
4. Pāriet vienu līmeni augstāk:
    ```bash
    cd ..
    ```
5. Izdrukāt mapes nosaukumu:
    ```bash
    pwd
    ```
6. Uztaisīt jaunu mapi savā mapē, nepārvietojoties uz to:
    ```bash
    mkdir janis_berzins/new_folder
    ```
7. Izdrukāt admina (savu) lietotāja vārdu:
    ```bash
    whoami
    ```
8. Pārvietoties uz jaunu mapi, izmantojot absolūto ceļu:
    ```bash
    cd /home/[lietotāja_vārds]/janis_berzins/new_folder
    ```
9. Palaist komandu cd bez argumentiem. Kurā mapē Jūs nonācāt? Ko, jūsuprāt, dara cd komanda bez argumentiem?
10. Izmēģiniet arī, ko dara šāda komanda:
    ```bash
    cd -
    ```
11. Izveidojiet jaunus tukšus failus:
    ```bash
    touch file1 file2
    ```
12. Ierakstiet otrajā failā šīs dienas datumu:
    ```bash
    date > file2
    ```
13. Izdrukājiet abu failu saturu uz ekrāna:
    ```bash
    cat file*
    ```
14. Pārsauciet pirmo failu, bet otru failu pārvietojiet vienu mapi augstāk, pēc tam pārbaudiet rezultātu ar ls komandu:
    ```bash
    mv file1 file1_new_name
    mv file2 ..
    ls . ..
    mv file2 ..
    ls . ..
    ```
15. Papildiniet pirmo failu ar mapē esošo failu sarakstu:
    ```bash
    ls >> file1_new_name
    cat file1_new_name
    ```
16. Saskaitiet failu skaitu savā mapē ar šādām komandām:
    ```bash
    ls ~ > file_list.txt
    wc -l file_list.txt
    rm file_list.txt
    ```
17. Izdrukājiet pēdējās 20 komandas, ko esat izmantojuši:
    ```bash
    fc -l -20
    ```
18. Izpildiet kādu no šīm komandām vēlreiz:
    ```bash
    fc -s [komandas_numurs]
    ```
19. Kāpēc `rm file*` komanda ir bīstama, un kā var šo sevi pasargāt no kļūdām? Atbildi var
atrast šajā lapā: (https://linuxcommand.org/lc3_lts0050.php)[https://linuxcommand.org/lc3_lts0050.php]
20. Kāpēc komandrinda var noderēt, manipulējot ar failiem, ja to pašu var izdarīt grafiskajā
vidē, izmantojot peli? Atbildi var atrast šajā lapā: (https://linuxcommand.org/lc3_lts0050.php)[https://linuxcommand.org/lc3_lts0050.php]
